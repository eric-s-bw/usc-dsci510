## Class and object
* objects are instances of class
* `__init__`: initialization
* additional methods: introduce
* instance namespace
* role of self

```python
class Student: # class statement

    school = "USC" # class attribute

    # __init__ is the function used to initialize an instance of class Student
    # self refers to the instance being initialized
    # but self is not a keyword (like this in Java)

    def __init__(self, name, major):
        self.name = name # name in self.name is an instance attribute
        self.major = major # major is also an instance attribute

    def change_major(self, major):
        self.major = major

    # david.introduce(): python will call this method, binding self to david
    def introduce(self):
        print(f'I am {self.name}. I major in {self.major}.')

# david is an instance of Student class
david = Student('David', "CS")
john = Student('John', "ADS")
```

## Special methods
* Special methods are methods with predefined double-underscore names that Python automatically uses for built-in operations and syntax.
* They are also called **dunder** methods.
* Examples:
    * `__repr__`: repr()
    * `__str__`: str() or print()
    * `__len__`: len()
    * `__contains__`: x in Student
    * `__eq__`: david == john
    * `__lt__`: david < john
    * `__call__`: david()

```python
class Student: # class statement

    school = "USC" # class attribute

    # __init__ is the function used to initialize an instance of class Student
    # self refers to the instance being initialized
    # but self is not a keyword (like this in Java)

    def __init__(self, name, major):
        self.name = name # name in self.name is an instance attribute
        self.major = major # major is also an instance attribute
        self.courses = []

    def add_course(self, course):
        self.courses.append(course)

    def change_major(self, major):
        self.major = major

    # david.introduce(): python will call this method, binding self to david
    def introduce(self):
        print(f'I am {self.name}. I major in {self.major}.')

    def __repr__(self): # programmer, debugging
        return f'Student(name={self.name}, major={self.major}, courses={self.courses})'

    def __str__(self):
        return f'({self.name}, {self.major}, {self.courses})'

    def __len__(self):
        return len(self.courses)

    def __contains__(self, course): # assume course is a string
        if not isinstance(course, str):
            raise TypeError('course is not a string')

        return course in self.courses

    # david == john
    # david.__eq__(john)
    def __eq__(self, other):
        if not isinstance(other, Student):
            raise TypeError(f'{other} is not an instance of Student')

        # print('inside eq')

        return (self.name == other.name and
                self.major == other.major and
                self.courses == other.courses)

    def __call__(self, x):
        print(x)
        self.introduce()

    # david < john
    def __lt__(self, other):
        if not isinstance(other, Student):
            raise TypeError(f'{other} is not an instance of Student')

        return self.name < other.name

# david is an instance of Student class
david = Student('David', "CS")
john = Student('John', "ADS")
```
