## Dynamic arrays
- list is store as a dynamic array

```python
lst = [1, 'hello', 3]

# storage structure of lst:
# length = 3
# capacity : 8
# [ ref1, ref2, ref3]
# [8, 8, 8]  # 8 bytes each
#  ref1 -> 1
#  ref2 -> 'hello'
#  ref3 -> 3

# assuem ref is 8 bytes
```

```python
id(1), id('hello'), id(3)
```

```python
# O(1)  const amount of time
len(lst)
```

```python
# O(n) for searching / linear search
lst.index(3)
```

## Linear search

```python
def linear_search(lst, x):
    for i in range(len(lst)):
        if lst[i] == x:
            return i
    return None

lst = [1, 2, 3, 4, 5]

linear_search(lst, 4)
```

## Binary search

```python
def binary_search(lst, x):
    left = 0
    right = len(lst) - 1

    while left <= right:
        # mid  = (left + right) // 2
        mid = left + (right - left) // 2
        if lst[mid] > x:
            right = mid - 1
        elif lst[mid] < x:
            left = mid + 1
        else:
            return mid
    return None

lst = [1, 2, 3, 4, 5]

linear_search(lst, 4)

linear_search(lst, 1)
linear_search(lst, 2)
```

## Insertion sort

```python
## original list [1, 2, 5, 4, 3, 8]
#  [1] [2, 5, 4, 3, 8] isolated the first vlaue
#  step1:  [1] [2] [5, 4, 3, 8]
#  step2:   1 < 2 -> do not change position [1, 2] [5, 4, 3, 8] -> [1, 2] [5] [4, 3, 8]
#  stpe3: [1, 2, 5] [4, 3, 8] -> [1, 2, 5] [4] [3, 8]-> 5 > 4 --> change position [1, 2, 4, 5] [3, 8]
#  stpe4: continue
#  step5: [1, 2, 3, 4, 5, 8]

def inserting_sort(lst):
    for i in range(1, len(lst)):
        curr = lst[i]
        j = i - 1

        while j >= 0 and lst[j] > curr:
            lst[j + 1] = lst[j]
            j -= 1
        lst[j + 1] = curr

    return lst

inserting_sort([1, 2, 5, 4, 3, 8])

```
