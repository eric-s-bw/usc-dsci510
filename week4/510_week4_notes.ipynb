## Dynamic arrays
* list is stored as a dynamic array

```python
lst = [1, 'hello', 3]

# storage structure of lst:
#  length = 3
#  capacity = 8
#  [ref1] [ref2] [ref3] [] [] [] [] [] <-- array of slots each storing a reference
#  new array: [r1] [r2] [r3] ... [r8] [r9] [] [] ... []
#  ^ <-- start address
#  [ 8      8     8]
#    ref1 -> 1
#    ref2 -> 'hello'
#    ref3 -> 3

# assuem ref is 8 bytes
```

```python
id(1), id('hello'), id(3)
```

```python
x = [1, 2]
lst = [1, 'hello', x]

# before pop, x and lst[2] refer to the same (nested) list: [1,2]

y = lst.pop()

# pop:
# 1. removes the last reference from lst
# 2. returns that reference
# 3. assigns that reference to y

# note pop does not destroy the nested list; it removes the reference from the
# outer list and returns the same reference

lst, x is y # True
```

## List operations and algorithmic complexities
* Complexity describes how the amount of work grows as the input size grows
  * We mostly focus on worst-case complexity
* n = input size
* O(1) operations:
  * len(lst)
  * lst[i]
  * lst.pop()
* O(k) operations:
  * lst[i:j] containing k elements
* O(n) operations:
  * x in lst
  * lst.index(x)
  * lst.pop(0): need to shift left
  * lst.insert(0, x): need to shift right
* Amortized O(1) operations:
  * lst.append(x)
* Amotized O(k) operations:
  * lst.extend(other) where other contains k elements

```python
lst = [1, 'hello', 3]
```

```python
# O(1): const amount of time
# the same amount of work is performed whether
# the list contains 3 or 3 million elements
```

```python
# len(lst): O(1)
# it just looks up the stored length

len(lst)
```

```python
# Indexing: O(1)
i = 2
lst[i]

# it can compute the address for lst[i] directly from base address
# and size of slot (each containing a reference of the same size, e.g., 8 bytes)
# = base address + i * size of reference
```

```python
# Slicing: O(k)

# slicing creates a new list containing k references
# (copied from the original list)

lst[1:3] # k elements selected and copied, k = 2 here
lst[:] # k = len(lst) = n here
```

```python
# Membership testing: O(n)
# may need to compare x with every element in the worst case
3 in lst
```

```python
# Searching a value in the list: O(n)
lst.index(3) # find the first index where 3 occurs in the list
```

```python
# Removing the last element: O(1)
# 1. just need to locate the last slot and remove (and return)
# the reference stored there: O(1)
# 2. length -= 1: O(1)
lst.pop() # lst.pop(len(lst) - 1)
```

```python
# Insertion near the beginning of list: O(n)
# 1. require shifting existing references to the right: O(n)
# 2. length += 1: O(1)
x = 4
lst.insert(0, x) # prepend
```

```python
# Removing the first element: O(n)
# 1. require shifting existing references to the left: O(n)
# 2. length -= 1: O(1)
lst.pop(0) # pop(i)
```

```python
lst = [1, 2, 3]
lst.append(4)
lst
```

## Amortization:
* some operations are usually inexpensive but occasionally expensive e.g.,
  * lst.append(x)
  * lst.extend(other)
* so we compute its amortized cost, which is the average cost per operation over a long sequence

```python
# Appending one element to a list: amortized O(1)
#
lst = [3, 2, 1]
lst.append(4)
lst

# Simplifying assumptions:
#   - Start with an empty list.
#   - Perform n appends, producing a final length of n.
#   - Capacity doubles whenever the backing array is full.
#   - n is a power of 2.

#  [ref1] [ref2] [ref3] [] [] [] [] [ref8] <-- array of slots each storing a reference
#  new array: [r1] [r2] [r3] ... [r8] [r9] [] [] ... []

# Case A: The backing array has an unused slot.
#   1. Store one reference in the first unused slot.
#   2. Increase the stored length by 1.
#
# Cost: O(1)

# Case B: The backing array is full.
# Suppose its current length is j, where j < n.
#
#   1. Allocate a larger backing array.
#   2. Copy the existing j references.
#   3. Store the new reference.
#   4. Increase the stored length by 1.
#
# Cost of this particular append: O(j).

# Under the simplified doubling model, the largest resize before
# reaching n elements occurs when j = n/2. That resize costs:
#
#   O(n/2) = O(n)

# Because capacity grows geometrically, resizing does not occur
# on every append. The cost of occasional resizing is amortized
# across the sequence of appends.

# Across n appends, the total number of references copied during
# resizing is:
#  S = 1 + 2 + 4 + ... + n/2 (geometric series)
#  = n - 1
#  = O(n)

# The n ordinary reference insertions also cost:
#
#   n * O(1) = O(n)

# Therefore, the total cost of n appends is:
#
#   O(n) + O(n) = O(n)

# Amortized cost per append:
#
#   O(n) / n = O(1)
```

```python
# Extend lst by k elements: amortized O(k)
#
lst = [3, 2, 1]
lst.extend([4, 7, 6])
lst

# Conceptually, we can think of extend as performing k appends.
# Each append has amortized O(1) cost.
#
# Therefore:
#
#   k * amortized O(1) = amortized O(k)

# More precisely, let:
#   j = the original length of lst
#   k = the number of elements added
#
# Case A: Enough unused capacity exists.
#   Store k new references.
#
#   Cost: O(k)
#
# Case B: Resizing is required.
#   1. Allocate a larger backing array.
#   2. Copy the j existing references.
#   3. Store the k new references.
#
#   Cost of this particular call: O(j + k)
#
# Across the list's sequence of additions, the occasional O(j)
# resizing work is amortized over the elements added during the
# list's growth history.
#
# Therefore, extending by k elements has amortized cost O(k).
```

```python
lst = [1, 2, 3]
lst.extend([4, 5])
lst
```

### Derivations

```python
# Simplified model: capacity doubles whenever the array is full.
# Assume n is a power of 2.

# Start with an empty list:
# capacity 0:  []

# Append the first element:
# capacity 1:  [ref1]

# Next growth copies 1 existing reference:
# capacity 2:  [ref1] [ref2]

# Next growth copies 2 existing references:
# capacity 4:  [ref1] [ref2] [ref3] []

# Next growth copies 4 existing references:
# capacity 8:  [ref1] [ref2] [ref3] [ref4] [ref5] [ ] [ ] [ ]

# ...

# The last resize on the way to n elements copies n/2 references.
# It increases capacity from n/2 to n.
#
# At that point, the list contains n/2 existing elements.
# The append that triggers resizing adds one new element.
# Subsequent appends fill the remaining capacity until the list
# contains n elements.

# Total resize-copying work:
# S = 1 + 2 + 4 + ... + n/2

# Multiply by 2:
# 2S = 2 + 4 + 8 + ... + n

# Subtract the first equation from the second:
# 2S - S = n - 1
# S = n - 1
# S = O(n)

# Each of the n appends also stores one new reference:
# n * O(1) = O(n)

# Total work for n appends:
# O(n) ordinary stores + O(n) resize copying = O(n)

# Amortized cost per append:
# O(n) / n = O(1)
```

## Linear search
* does not require sorted data
* complexity
  * best case: O(1)
  * worst case: O(n)

```python
def linear_search(lst, x):
  for i in range(len(lst)):
      if lst[i] == x:
          return i

  return None

lst = [3, 6, 1, 5, 2, 5, 8, 9, 6, 7]
linear_search(lst, 5)
```

## Binary search
* require sorted data
* repeatedly discard half of the remaining candidates
* complexity: $O(\log n)$

```python
def binary_search(lst, x):
    low = 0
    high = len(lst) - 1

    while low <= high:
        mid = (low + high) // 2

        if x < lst[mid]:
            high = mid - 1
        elif x > lst[mid]:
            low = mid + 1
        else:
            return mid

    return None


lst = [3, 6, 1, 5, 2, 7, 4]
lst.sort()
print(lst)
binary_search(lst, 5)
```

* Search space:
    * $n \rightarrow \frac{n}{2} \rightarrow \frac{n}{4} \rightarrow ... \rightarrow 1$

* Number of halving/division:
    * $n/2^k = 1$
    * $k = \log_2 n$
* Therefore: complexity is $O(\log n)$

## Insertion sort
* Complexity: $O(n^2)$

```python
lst = [3, 6, 1, 5, 2, 5, 8]
#     [3]        [6, 1 ..., 2, 5, 8 ]
#     ^sorted          ^unsorted
#
#     [3] 6 <--- insert first element in unsorted into sorted
#     [3, 6] 1
#     [1, 3, 6] 5
#     [1, 3, 5, 6] 2
#     [1, 2, 3, 5, 6] 5
#     [1, 2, 3, 5, 5, 6] 8
#     [1, 2, 3, 5, 5, 6, 8]
```

```python
def insertion_sort(lst):
    for i in range(1, len(lst)): # lst[:i] is sorted portion, lst[i:] unsorted
        x = lst[i]
        pos = i

        for j in range(i-1, -1, -1):
            if lst[j] > x:
                lst[j+1] = lst[j]
                pos = j
            else:
                break

        lst[pos] = x

lst = [3, 6, 1, 5, 2, 5, 8]
insertion_sort(lst)
lst
```

```python
# Complexity
# 1 + 2 + ... + (n-1) = n(n-1)/2 = O(n^2)
```
