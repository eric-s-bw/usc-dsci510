### Indexing

```python
#lst = []
lst = [3, 2, 5, 1, 5, -5]

#      0  1   ...      5   // 0 to len(lst) - 1
#     -6              -1   // -len(lst) to -1

# -len(lst)  to len(lst) - 1

#len(lst)
```

```python
# index starts from zero
lst[0], lst[2]
# negative index
lst[-1] # refers to last in list
lst[-2], lst[-6]

# lst[-7] # out of range
# lst[6] # IndexError: out of range
```

```python
# a valid negative index can be converted by adding the length

# -i is a negative index
# lst[-i]  == lst[len(lst) - i]

lst[-1] == lst[6 - 1] == lst[5]
lst[-6] == lst[6 - 6] == lst[0]
```

### Slicing

```python
lst, len(lst)
```

```python
# slicing a list produces a new list

# lst[start:stop:step]
# stop is exclusive

lst[0:] == lst[0:len(lst):1] # default step = 1, default stop is len(lst)

# get the first two elements of lst
lst[0:2] == lst[:2] == lst[0:2:1] # default step = 1, default start = 0

lst[:] == lst[0:len(lst)] == lst[0:len(lst):1]
```

```python
lst[1:5] # lst[1] ... lst[4]
range(1, 5) # 1, 2, ..., 4
```

```python
range(5) # 5 = stop
range(1, 5) # default step = 1
list(range(1, 5)) == list(range(1, 5, 1))
list(range(5)) == list(range(0, 5))
```

```python
# negative start/stop indices count from the end
lst[-1:] == lst[5:] # start = -1 + len(lst) = -1 + 6 = 5

# how to get a list with only the last two
lst[-2:] == lst[4:] == lst[4:6] == lst[-2:6] == lst[-2:6:1]
```

```python
# slicing is flexible with (out of range) index
lst[-100:] == lst[-7:] == lst[-6:]
```

```python
# for a negative step, default start = last in list,
# default stop = pass through first ln list

lst, lst[::], lst[::-1]
```

```python
lst[5::-1], lst[5:0:-1], lst[5:-1:-1] == lst[5:5:-1]
```

```python
lst[-3:-1:] == lst[3:5:1]
```

```python
lst[-3:-1:-1] == lst[3:5:-1]
```

```python
lst[3:5:-1]
```

```python
lst, len(lst)
```

```python
list(range(5, -1, -1)), lst[5:-1:-1]
```

```python
lst
lst[::2]  == [-3, -5, 5]
```

```python
# explicit negative stop vs omitted stop
lst = [1, 2, 3, 4, 5]
print(lst[5:-1:-1])   # explicit -1 → convert to index 5 → []
lst[5::-1]     # stop omitted → continue through index 0
```

### Pass lists to functions

```python
lst1 = [-5, 1, 2, 3, 5, 5]

def f(lst): # lst is a parameter
    lst[0] = -100 # this mutates lst
    # what about this?
    #lst.append(6)

print(lst1)

f(lst1) # lst = lst1 (parameter lst is bound to the same object as lst1)

print(lst1)
```

```python
lst1 = [-5, 1, 2, 3, 5, 5]

def f(lst):
    lst = []

print(lst1)
f(lst1) # lst = lst1, then lst = [] (lst is rebound to [])
print(lst1)
```

```python
def f(y):
    y = 3
    print(y)

x = 5
f(x) # # y is initially bound to the same object as x;
       # then y is rebound to 3
x
```

## Iterables, iterators, generators
* iterable: an object that can provide an iterator (using iter())
    * e.g., lst is an iterable and iter(lst) an iterator
* iterator:
    * represents a stream of items
    * to get its next item, use next()
    * note an iterator is also an iterable
        * iter(it) is it
* generator: a special kind of iterator generated by:
    * generator expression: (x**2 for x in range(5))
    * generator function: using yield


```python
lst = [1, 2, 3, 4, 3, 2, 1]
```

```python
for x in lst:
    print(x)
```

```python
# what is behind the above for loop?

it = iter(lst) # lst is an iterable, it is its iterator

while True:
    try:
        x = next(it)
    except StopIteration:
        break

    print(x)
```

```python
iter(it) is it
```

```python
# (x for x in range(5)) is a generator expression
# g1 and g2 is a generator object

# evaluating the expression ==> a generator object

g1 = (x for x in range(5))
g2 = (x for x in range(5))
g1 is g2
```

```python
list(g1)
```

```python
# generator function

# note yield lets a generator function produce a value,
# pause its execution state, and resume on the next request.
def square(n):
    for x in range(n):
        yield x ** 2

g2 = square(3)
next(g2), next(g2), next(g2)
```

## Exceptions
* exceptions are objects used to signal exceptional conditions during execution
* common examples: ValueError, ZeroDivisionError, IndexError, KeyError, StopIteration


```python
# examples
# int("abc")     # ValueError
# 10 / 0         # ZeroDivisionError
# lst[100]       # IndexError
# student["age"] # KeyError
```

```python
try:
    x = int("abc")
except ValueError:
    print("Invalid integer")
```

```python
#Execution leaves the `try` block when an exception occurs
try:
    print('A')
    x = int('abc')
    print('B')
except ValueError:
    print('C')

print('D')
```

```python
# Different exception types can have different handlers
try:
    x = int(input('x: '))
    y = int(input('y: '))
    print(x / y)
except ValueError:
    print('Invalid integer')
except ZeroDivisionError:
    print('Cannot divide by zero')
```

```python
# Exceptions are objects
try:
    int('abc')
except ValueError as e:
    print(type(e))
    print(e)

    # isinstance(e, ValueError)
    # True
```

```python
# `else` runs when no exception occurs
try:
    x = int(input('Enter an integer: '))
except ValueError:
    print('Invalid input')
else:
    print('x =', x)
```

```python
#`finally` runs regardless
try:
    x = int(input('Enter an integer: '))
except ValueError:
    print('Invalid input')
finally:
    print('Done')
```

```python
# `raise` explicitly signals an exception
def average(scores):
    if len(scores) == 0:
        raise ValueError('scores cannot be empty')

    return sum(scores) / len(scores)

#average([])
# ValueError
```
