### range

```python
for x in range(1, 5, 2): # start is inclusive, stop is exclusive
    print(x)
```

```python
help(range)
```

```python
range(5) # stop = 5
range(0, 5) # start = 0, stop = 5
range(0, 5, 1) # start = 0, stop = 5, step = 1
```

```python
for x in range(5, 1, 1):
    print(x)

# x = start
# if x < 1:
#    print(x)

# or more completely
start = 5
stop = 1
step = 1

x = start
while x < stop: # note for positive step, x approaches stop from left
    print(x)
    x += step
```

```python
for x in range(5, 1, -1):
    print(x)

# x = start
# if x > stop:
#   print(x) <--- 5
#   x = start + step = 5 - 1 = 4
#   ...
#   print(x)

# or more completely
start = 5
stop = 1
step = -1

x = start
while x > stop: # note for negative step, x approaching stop from right
    print(x)
    x += step
```

```python
for x in range(5, 0, -1):
    print(x)
```

```python
for x in range(5, 1, -1): # 1 is also exclusive
    print(x)

# x = 5
# if x > 1:
#   print(x)
```

A useful rule:
* for positive step, start < stop
* for negative step, start > stop

otherwise, range is empty

```python
# also step can not be zero

#this will raise error
#range(1, 5, 0)
```

### break vs continue
* break: out of the nearest loop
* continue: to the next iteration, i.e.,skip the remainder of the current iteration

```python
for x in range(5):
    print('x =', x)
    for y in range(x + 1, 5):
        if y == 3:
            break
        print('\ty =', y)
```

```python
for x in range(1, 6):
    if x == 3:
        continue
    print(x)
```

```python
for x in range(1, 6):
    if x == 3:
        break
    print(x)
```
