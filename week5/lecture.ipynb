# Hashing
- Hashing converts a hashable object into an integer.
- An dictionary or set use that integer to begin locating the object.
- Equal objects must habe equal hashes. 
- Unequal objects may havethe same hash (collision).

```python
lst = ['johmn', 'bill', 'mary']

'bill' in lst # complexity O(n)

# hashing idea today will allow us to do look up in O(1) time
```

```python
1 == 1.0, hash(1) == hash(1.0), hash(1), hash(1.0)
```

```python
hash(-1), hash(-2), -1 == -2, hash(-1) == hash(-2) # collision
```

```python
# maybe hashable
# tuple is hashable only when its elements are all hashable
hash((1, 2)) # yes
# hash((1, [2])) # no - lists are not hashable 
```

```python
# unhashable: list, dict, set
# hash([])  # an empty list
# hash({}) # an empty dict
# hash(set()) # an empty set
```

```python
# hashables:
#   int, float, str, bool, (some)tuple, NoneType, range
hash(1), hash(1.2), hash('john'), hash(True), hash(False), hash((1, 2)), hash(None), hash(range(10))
```

# Hash table
- Hash tables use hashing and collision resolution to provide expected O(1) lookup, insertion, and deletion.
  - linear probing 

```python
# hash table:
capacity = 7

# hashing name
# hash(name) % capacity --> first position to lookup

hash("john") % capacity # 2
# 2 is free, put john in position 2
hash('david') % capacity # 2
# 2 is occupied, try pos + 1 == 5
# 5 is free, put david in position 5

# Index:   0     1     2     3     4     5     6
# Value:             'john'
# 
```

```python
hash('john'), hash('john') % capacity, (hash('john') + 1) % capacity, (2 + 1) % capacity
```

```python
a = {1, 2, 3, 4}
b = {2, 3, 4, 5, 6, 7, 8, 9}
```

```python
def intersection(a, b):

    result = set()
    for x in a:
        if x in b: # expected O(1) time
            result.add(x)
    return result

```
