## Inheritance
- object is the root class (base class) in the class inheritance hierarchy
  - `__bases__`
- every class is a subclass of object
  - `subclass()`

```python
class Student: # same as class Student(object):
    school = 'USC' # class attribute
    
    def __init__(self, name, major): # constructor
        self.name = name
        self.major = major
    
    def introduce(self):
        print(f'I am {self.name} and I major in {self.major} at {self.school}.')

# david is an instance of Student class
david = Student('David', 'CS') # call __init__
david

john = Student('John', 'ADS')

# both __init__ and introduce are instance methods
```

```python
issubclass(Student, object)
```

```python
object.__bases__ #  return an empty tuple because object is the root class
# e.g., object is the base class of Student
```

```python
Student.__bases__ # base classes which are direct superclass 
```

### super()

```python
class TA(Student):
    def __init__(self, name, major, hours):
        self.name = name
        self.major = major
        self.hours = hours
    
    

mary = TA('Mary', 'Math', 20)
mary.introduce()
```

```python
# TA class overrides __init__ and introduce
class TA(Student):
    def __init__(self, name, major, hours):
        self.name = name # better done by Student
        self.major = major # this too
        self.hours = hours # 

    def introduce(self):
        print(f'I am {self.name} and I major in {self.major} at {self.school}.') # Student does it
        print(f'I work {self.hours} hours per week.')

mary = TA('Mary', 'Math', 20)
mary.introduce()
```

```python
class Employee:
    pass

class TA(Student, Employee):
    def __init__(self, name, major, hours): 
        # super().__init__(name, major)

        super(TA, self).__init__(name, major)

        # conceputally,
        # Student.__init__(self, name, major)

        self.hours = hours

    def introduce(self):
        super().introduce()
        print(f'I work {self.hours} hours per week.')

mary = TA('Mary', 'Math', 20)        

isinstance(mary, TA), isinstance(mary, Student)

mary.__dict__
```

```python
TA.__bases__
```

```python
issubclass(TA, object) # note subclass relationship is transitive
```

```python
for base in TA.__bases__:
    print(base, base.__bases__)

# __bases__ only give you direct superclasses
# you can write a (recursive/iterative) function to find all superclasses
```
