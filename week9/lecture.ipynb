## Instance and class atterbutes, namespaces and shadowing

```python
class Student:
    count = 0 # class attribute

    def __init__(self, name):
        self.name = name # instance attribute
        self.count += 1
        print('self.__dict__:', self.__dict__)
        print('self.count:', self.count)

david = Student('David')
# john = Student('John')
Student.count , david.count
```

```python
'count' in Student.__dict__, 'count' in david.__dict__
```

## instance methods

```python
class Student:
    def __init__(self, name, major):
        self.name = name
        self.major = major

    def introduce(self): # instance method, self refers to the instance
        print('slef = ', self)
        print(f'I am {self.name}. I am majoring in {self.major}.')

    @classmethod # decorator
    def from_string(cls, text):
        print('cls=', cls)
        name, major = text.split(',')
        return cls(name, major)

    @staticmethod # utility function
    def valid_major(major):
        return major in {'CS', 'ADS', 'Math'}

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        if not value:
            raise ValueError('Name cannot be empty.')
        self._name = value



# bill= Student('Bill', 'DSCI')
# bill.introduce()

david = Student.from_string('David, DSCI')
david.introduce()
```
