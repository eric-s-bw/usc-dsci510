## Sequences: ordered, can be indexed or sliced
- list : mutable
- tuple : immutable
- str : immutable
- range : immutable

```python
lst = [1, 2, 3]
len(lst), lst[0], lst[:2]
```

```python
s = 'hello'
len(s), s[0], s[-1], s[-2:]
```

```python
# s[0] = 'H'  # TypeError: 'str' object does not support item assignment
# strings are immutable
```

```python
r = range(5)
len(r), r[0], r[-1], r[:2]
```

## Dynamic arrays
- list is store as a dynamic array

```python
lst = [1, 'hello', 3]

# storage structure of lst:
# length = 3
# capacity : 8
# [ ref1, ref2, ref3]
# [8, 8, 8]  # 8 bytes each
#  ref1 -> 1
#  ref2 -> 'hello'
#  ref3 -> 3

# assuem ref is 8 bytes
```

```python
id(1), id('hello'), id(3)
```

```python
# O(1)  const amount of time
len(lst)
```

```python
# O(1)
lst[2]

# address for lst[2]
# = base address + 2 * size_of_ref
```

```python
# O（k)
lst[1:3] # k = 2, number of elements in the slice
```

```python
# O(n) for searching / linear search
lst.index(3)
```
