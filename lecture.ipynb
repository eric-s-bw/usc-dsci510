## Lists

```python
# a ordered mutable sequence of data

lst = []
lst = [3, 2, 5, 1, 5, -5]
#      0, 1 ....       5 // 0 to len(lst) - 1
#     -6, -5 ....     -1  // -len(lst) to -1
#     -len(lst) to len(lst) - 1
```

## Indexing

```python
# index starts from zero
lst[0]

# negative index
lst[-1] # refer to last in list
lst[-2]


# lst[-7] # IndexError: list index out of range

# a valid negative index can be converted by adding len(lst) to it
# i is a negative index
# lst[-i] == lst[len(lst) - i]

lst[-1] == lst[len(lst) - 1] == lst[5]
lst[-6] == lst[len(lst) - 6] == lst[0]
```

## Slicing

```python
lst
```

```python
# lst[start:stop:step]

# first try
lst[0:] == lst[0:len(lst):1]

# get the first two elements of lst
lst[0:2]

lst[0:2] == lst[:2] == lst[0:2:1] # default step = 1, default start = 0

lst[:] == lst[0:len(lst)] == lst[0:len(lst):1]
```

```python
# negative start/stop indices count from the end
lst[-1:] == lst[5:] # start = -1 + len(lst) = 5

lst[-2:] == lst[4:] == lst[4:6] == lst[-2:6] == lst[-2:6:1]
```

```python
# slicing is flexible with (out of range) index

lst[-7:] == lst[:]

lst[-6:]

lst[-100:] == lst[-7:] == lst[-6:]
```

```python
# for a negative step, default start = last in list
# default stop = pass through first in list

lst, lst[::],lst[::-1]
```

```python
lst[5::-1]
```

```python
lst[:1:-1]
```

```python
list(range(5, -1, -1)), lst[5:-1:-1]
```

## identity test vs content equality test

```python
# == testing equality of content/value
lst == lst[:] == lst[::]
lst, lst[:]
```

```python
# is tests identity  slicing creates a new list
lst is lst[:], id(lst), id(lst[:])
```

## Mutablity
- methods:
  - insert(index, value)
  - append (x)
  - extend([x....])
  - x = pop(i)
  - remove(x)

```python
# python list is mutable; its content may be changed
print(lst)
lst[0] = -3
print(lst)
```

```python
# a slice can also be assigened to 
lst[1:3] = [-2, -5]
```

```python
# methods
# both append and extend change the content of orignial list
#

lst.append(0)
lst
```

```python
lst.extend([10,20]) # change the content of lst
lst
```

```python
lst
```

```python
lst.insert(0, -10)
lst
```

```python
# queue: FIFO
# stack: FILO

# queue: append (add to the end)
lst = [1, 2, 3]
lst.append(4) # into queue
x = lst.pop(0) # 1 is the first element in the list, pop(0) removes it from the list  // out of queue
x, lst
```

```python
y = lst.pop() # 4 is the last element in the list, pop() removes it from the list  // out of stack
y, lst
```

```python
# remove the first occurrence of a value in the list
lst = [1, 2, 2, 3]
lst.remove(2)
lst
# lst.remove(4) # ValueError: list.remove(x): x not in list
```

```python
# remove the second element from the list by index
del lst[1] 
lst
```
