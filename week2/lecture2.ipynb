## break vs continue
- break: out of the nearest loop
- continue: to the next iteration, i.e.,skip the remainder of the current iteration
- pass: doing nothing

```python
for x in range(5):
    print('x = ', x)

    for y in range(x + 1, 5):
        if y == 3:
            break
        print('\ty = ', y)


```

```python
for x in range(5):
    print('x = ', x)

    for y in range(x + 1, 5):
        if y == 3:
            continue
        print('\ty = ', y)
```

```python
lst = [3, 4, 6, -2, 5, 0, 9]

x = 0 
sum = 0

for x in lst:
    if x < 0:
        break
    sum += x

print(sum)
```

```python
lst = [3, 4, 6, -2, 5, 0, 9]

x = 0 
sum = 0

for x in lst:
    if x < 0:
        continue
    sum += x

print(sum)
```

```python
lst = [3, 4, 6, -2, 5, 0, 9]

x = 0 
sum = 0

for x in lst:
    if x < 0:
        pass
    sum += x

print(sum)
```

```python
def double(x): # x is a parameter, function name is double
    return x * 2

d = double(2) # 2 is an argument of function call
# 2 --- > x
# x = 2
# return 2 * 2
print(d)
```

```python
def add(x, y): # 2 parameters
    return x + y

z = add(3, 5) # 2 positional arguments
# 3 --- > x
# 5 --- > y
# x = 3, y = 5
# return 3 + 5
# print(z)

z = add(y=3, x=5) # 2 keyword arguments
```

## Scope
- LEGB:
  - L: local
  - E: enclosing
  - G: global
  - B: build-in

```python
def f(x, y):
      z = 3
      return x + y + z

# x, y, z are all local variables

f(1, 2)
```

```python
z = 3 # global variable

def f(x, y):
      return x + y + z

f(1, 2)
```

```python
z = 3 # global variable

def f(x, y):
      z = 4 # local variable, shadowing
      return x + y + z

f(1, 2), z
```

```python
z = 3 # global variable

def f(x, y): # x, y are local parameters
      global z # global variable, no shadowing
      z = 4 # change global variable z
      return x + y + z

f(1, 2), z
```

```python
# E: enclosing

x = 4

def outer():
    x = 5

    def inner():
        nonlocal x # change the value of x in the enclosing scope (outer function)
        x = 6
        print('inner x = ', x)

    inner()
    print('outer x = ', x)

outer()
print('global x = ', x)

# L E G B 
```

## Recursion:
- a function that calls itself

```python
# 1 + 2 + 3 + .. + n

#iterative version
def sum_to(n):
    sum = 0
    for x in range(1, n + 1): # range(n, 0, -1)
        sum += x
    return sum

print(sum_to(10))
```

```python
# write recursion function
# base case:
#    sum_to(1) = 1
# recursive case:
#    sum_to(n) = n + sum_to(n - 1) for n > 1


def sum_to_rec(n): # n >= 1
    if n == 1:
        return 1
    # recursive case: sum_to_rec(n - 1)
    return n + sum_to_rec(n - 1)

print(sum_to_rec(10))

sum_to(3) == sum_to_rec(3)
```

```python
# sum_to_rec(3)
# hit the base, then backtrack

# => 3 + sum_to_rec(2) => 3 + 3 => 6
# => sum_to_rec(2) => 2 + sum_to_rec(1) => 2 + 1 => 3
# => sum_to_rec(1) => 1 (base case)
```

```python
# factorial
# 3! = 3 * 2 * 1 = 6

def fac(n):
    res = 1

    for x in range(1, n+1):
        res *= x
    return res

print(fac(3))
```

```python
# base case: fac_rec(0) = 1
# recursive case: fac_rec(n) = n * fac_rec(n - 1) for n > 0



def fac_rec(n): # n >= 0
    if n == 0:
        return 1
    return n * fac_rec(n-1)

print(fac_rec(3))
```

## Under the hood

- dis

```python
# interpreter
# code -> bytecode -> PVM (python virtual machine)
g = 5

def f(x): 
    y = 3
    return x + y + g + 2

```
