## break vs continue
- break: out of the nearest loop
- continue: to the next iteration, i.e.,skip the remainder of the current iteration
- pass: doing nothing

```python
for x in range(5):
    print('x = ', x)

    for y in range(x + 1, 5):
        if y == 3:
            break
        print('\ty = ', y)


```

```python
for x in range(5):
    print('x = ', x)

    for y in range(x + 1, 5):
        if y == 3:
            continue
        print('\ty = ', y)
```

```python
lst = [3, 4, 6, -2, 5, 0, 9]

x = 0 
sum = 0

for x in lst:
    if x < 0:
        break
    sum += x

print(sum)
```

```python
lst = [3, 4, 6, -2, 5, 0, 9]

x = 0 
sum = 0

for x in lst:
    if x < 0:
        continue
    sum += x

print(sum)
```

```python
lst = [3, 4, 6, -2, 5, 0, 9]

x = 0 
sum = 0

for x in lst:
    if x < 0:
        pass
    sum += x

print(sum)
```

## Functions， parameters, and arguments
- function definition, e.g., def f(x, y)
  vs function call, .e.g., f(3, 5)
  - parameters: 
    - no parameter, single, multiple paremeters
    - parameter can take default value
      - no-default -> default
  - arguments:
    - positional -> keyword

```python
%%weitefile get_score.py
```

```python
# function definition
def get_score(): #header
    score = int(input('enter a score: '))
    return score

s = get_score() # assignment statement
print('score =', s)
```

```python
def double(x): # x is a parameter, function name is double
    return x * 2

d = double(2) # 2 is an argument of function call
# 2 --- > x
# x = 2
# return 2 * 2
print(d)
```

```python
def add(x, y): # 2 parameters
    return x + y

z = add(3, 5) # 2 positional arguments
# 3 --- > x
# 5 --- > y
# x = 3, y = 5
# return 3 + 5
# print(z)

z = add(y=3, x=5) # 2 keyword arguments
```

```python
def subtract(x, y):
    return x - y

subtract(3, 5) == subtract(5, 3) # False
subtract(x=3, y=5) == subtract(y=5, x=3) # True
```

```python
subtract(3, y=5)
```

```python
subtract(x=3, 5) # invalid syntax
```

```python
def f(x, y = 0):
    return x + y

f(3)
f(3, 5)
```

```python
# rules
#   parameters: non-default (followed by) default, e.g., def f(x, y = 5)
#   arguments: positional (followed by) keyword, e.g., f(3, y = 5)
```

```python
def f(x, y):
    return x + y

f(3, 4)
```

## Scope
- LEGB:
  - L: local
  - E: enclosing
  - G: global
  - B: build-in

```python
def f(x, y):
      z = 3
      return x + y + z

# x, y, z are all local variables

f(1, 2)
```

```python
z = 3 # global variable

def f(x, y):
      return x + y + z

f(1, 2)
```

```python
z = 3 # global variable

def f(x, y):
      z = 4 # local variable, shadowing
      return x + y + z

f(1, 2), z
```

```python
z = 3 # global variable

def f(x, y): # x, y are local parameters
      global z # global variable, no shadowing
      z = 4 # change global variable z
      return x + y + z

f(1, 2), z
```

```python
# E: enclosing

x = 4

def outer():
    x = 5

    def inner():
        nonlocal x # change the value of x in the enclosing scope (outer function)
        x = 6
        print('inner x = ', x)

    inner()
    print('outer x = ', x)

outer()
print('global x = ', x)

# L E G B 
```
