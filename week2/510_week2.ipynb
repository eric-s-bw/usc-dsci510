## Functions, parameters, and arguments
* function definition, e.g., def f(x, y)
 vs function call, e.g., f(3, 5)
* parameters:
    * no parameter, single, multiple parameters
    * parameter can take default value
        * no-default -> default
* arguments:
    * positional -> keyword

```python
def double(x): # x is a parameter
    return x * 2

d = double(2) # 2 is an argument of function call
# 2 ---> x
# x = 2
print(d)
```

```python
def add(x, y): # 2 parameters
    return x + y

z = add(3, 5) # 2 arguments
# 3 --> x, 5 --> y
# x = 3, y = 5
print(z)
```

```python
def add(x, y): # 2 parameters
    return x + y

z = add(3, 5) # 2 positional arguments
# 3 --> x, 5 --> y
# x = 3, y = 5
#print(z)

#z = add(x = 3, y = 5)
z = add(y = 5, x = 3) # keyword arguments
print(z)
```

```python
def subtract(x, y):
    return x - y

subtract(3, 5) == subtract(5, 3)
#subtract(x = 3, y = 5) == subtract(y = 5, x = 3)
```

```python
# If you have both positional and keyword arguments,
#   use position first, then names.
# Once you start specifying arguments by parameter name,
# you cannot go back to ordinary positional arguments.
```

```python
subtract(3, y = 5) # positional followed by keyword arguments
```

```python
#subtract(x = 3, 5) # error because 5 (a positional argument)
# follows a keyword argument (x = 3)
```

```python
# parameters with default values
def f(x = 5, y = 0):
    return x + y

f(3), f(3, 0), f()
```

```python
# parameter y has a default value 0
def f(x, y = 0):
    return x + y

f(3), f(x=3) # use defaults
f(3, 5), f(3, y=5) # override defaults

#f(x=3, 1) # not ok (positional arguments need to come before keyword args)
```

```python
# parameters without defaults must precede parameters with defaults
def f(x = 5, y): # error!
    return x + y
```

## Scope
* LEGB:
    * L: local
    * E: enclosing
    * G: global
    * B: builtin

```python
def f(x, y):
    z = 3
    return x + y + z

# x, y, z are all local variables

f(1, 2)
#x, y, z
type(f)

# inside f call
#   x = 1
#   y = 2
#   z = 3
```

```python
z = 3 # global variable

def f(x, y):
    return x + y + z

f(1, 2)
z
```

```python
z = 3 # global variable

def f(x, y):
    z = 4 # local variable, shadowing
    return x + y + z

f(1, 2), z
```

```python
z = 3 # global variable

def f(x, y): # x, y are local parameters
    global z
    z = 4 # change global variable z
    return x + y + z

def g():
    print(z)

g()
```

```python
g()
```

```python
# E: enclosing

x = 4 # global

def outer():
    #x = 5 # enclosing

    def inner():
        #x = 6 # local
        print(x)

    inner()

outer()
```

```python
# E: enclosing

x = 4 # global

def outer():
    x = 5 # enclosing

    def inner():
        global x
        x = 6
        print('inner x =', x)

    inner()
    print('outer x =', x)

outer()
print('global x =', x)
```

## Recursion:
* a function that calls itself

```python
# 1 + 2 + 3 + ... + n

# iterative version
def sum_to(n):
    s = 0

    for x in range(1, n+1): # range(n, 0, -1):
        s += x

    return s

sum_to(10)
```

```python
# write recursive version of sum_to
# base case:
#   n = 1
# recursive case:
#   sum_to(n) = n + sum_to(n-1)

def sum_to_rec(n): # n >= 1
    if n == 1: # base case
        return 1

    # all codes here will be executed when n != 1 (n > 1)
    # recursive case
    return n + sum_to_rec(n - 1)


sum_to(3) == sum_to_rec(3)
```

```python
# sum_to_rec(3)
# hit the base, then backtrack

# => 3 + sum_to_rec(2) => 3 + 3 => 6
#        => 2 + sum_to_rec(1) => 2 + 1 => 3
#                => 1 (hit base)
```

```python
# factorial
# 3! == 3 * 2 * 1 == 6
```

```python
def fac(n):
    res = 1

    for x in range(1, n+1):
        res *= x

    return res
```

```python
fac(3) == 6
```

```python
# base case:
#    n = 0 => 1
# recursive case:
#    fac(n) = n * fac(n - 1)

def fac_rec(n): # n >= 0
    if n == 0:
        return 1

    return n * fac_rec(n - 1)

fac_rec(3)
```

```python
%%writefile mymath.py
print('inside mymath', __name__)

def sum_to(n): # n >= 1
    if n == 1:
        return 1

    return n + sum_to(n - 1)

def fac(n): # n >= 0
    if n == 0:
        return 1

    return n * fac(n - 1)
```

## Under the hood
* code -> bytecode -> PVM
* dis
    * co_varnames (parameters and local variables)
    * co_consts
    * co_names (global variables)
* compile
    * modes: "exec" (statement) and
      "eval" (expression)



```python
# what Python interpreter does to your code:
#   source code -> bytecode -> PVM (python virtual machine)
#   bytecodes are platform indepedent
```

```python
def f(x):
    y = 3
    return x + y + 2
```

```python
from dis import dis
dis(f) # or dis(f.__code__)
```

```python
# Python evaluates your code using an evaluation stack
# evaluation stack

# stack vs queue
#   FIFO: queue
#   FILO: stack

dis(f)
```

```python
# evaluation stack:
#  [3]
#  [] 3 -> y
#  [x, y]
#  [x+y, 2]
#  [x+y+2] # binary_op + pushes result onto stack
#  [] # pop the top of stack and return it (x + y + 2)
```

```python
# 1 byte = 8 bits
# 1111 0001 => LOAD_FAST
# 0000 0000 => 0 (index into co_varnames)
```

```python
f.__code__.co_varnames # parameters and local variables
# slots
```

```python
f.__code__.co_consts
```

```python
f.__code__.co_names # global variables
```
